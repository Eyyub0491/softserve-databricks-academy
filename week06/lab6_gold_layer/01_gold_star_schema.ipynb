# Lab 6 - Gold Star Schema

Creates the Gold layer tables used for analytics.

## Star Schema
- **Dimensions**: dim_date, dim_customers, dim_products
- **Fact**: fct_sales_orders (order line items, one row per product ordered)
- **Aggregations**: agg_customer_summary (customer lifetime value), agg_daily_sales (daily metrics for alerting)

## Source Data
- lab5.silver.slv_sales_orders_clean
- lab5.silver.slv_customers_clean

```python
%sql
CREATE OR REPLACE TABLE lab5.gold.dim_date
COMMENT 'Date dimension table covering sales order period (July-Dec 2019) with standard calendar attributes'
AS
WITH date_spine AS (
  SELECT 
    EXPLODE(
      SEQUENCE(
        TO_DATE('2019-07-01'), 
        TO_DATE('2019-12-31'), 
        INTERVAL 1 DAY
      )
    ) AS full_date
)
SELECT 
  CAST(DATE_FORMAT(full_date, 'yyyyMMdd') AS INT) AS date_key,
  
  full_date,
  
  YEAR(full_date) AS year,
  QUARTER(full_date) AS quarter,
  MONTH(full_date) AS month,
  DATE_FORMAT(full_date, 'MMMM') AS month_name,
  
  WEEKOFYEAR(full_date) AS week_of_year,
  DAYOFMONTH(full_date) AS day_of_month,
  DAYOFWEEK(full_date) AS day_of_week,
  DATE_FORMAT(full_date, 'EEEE') AS day_name,
  
  CASE WHEN DAYOFWEEK(full_date) IN (1, 7) THEN TRUE ELSE FALSE END AS is_weekend
  
FROM date_spine
ORDER BY full_date;
```

```python
%sql
SELECT 
  'Total rows' AS metric,
  CAST(COUNT(*) AS STRING) AS value
FROM lab5.gold.dim_date

UNION ALL

SELECT 
  'Min date' AS metric,
  CAST(MIN(full_date) AS STRING) AS value
FROM lab5.gold.dim_date

UNION ALL

SELECT 
  'Max date' AS metric,
  CAST(MAX(full_date) AS STRING) AS value
FROM lab5.gold.dim_date

UNION ALL

SELECT 
  'Min date_key' AS metric,
  CAST(MIN(date_key) AS STRING) AS value
FROM lab5.gold.dim_date

UNION ALL

SELECT 
  'Max date_key' AS metric,
  CAST(MAX(date_key) AS STRING) AS value
FROM lab5.gold.dim_date

UNION ALL

SELECT 
  'Weekday days' AS metric,
  CAST(COUNT(*) AS STRING) AS value
FROM lab5.gold.dim_date
WHERE is_weekend = FALSE

UNION ALL

SELECT 
  'Weekend days' AS metric,
  CAST(COUNT(*) AS STRING) AS value
FROM lab5.gold.dim_date
WHERE is_weekend = TRUE;
```

```python
%sql
CREATE OR REPLACE TABLE lab5.gold.dim_customers
COMMENT 'Current customer dimension with demographics, loyalty status, and order flags for Lab 6 Gold analytics'
AS
WITH current_customers AS (
  SELECT 
    customer_id,
    customer_name,
    tax_id,
    tax_code,
    state,
    city,
    region,
    postcode,
    loyalty_segment,
    units_purchased,
    valid_from_ts,
    ROW_NUMBER() OVER (
      PARTITION BY customer_id 
      ORDER BY valid_from_ts DESC
    ) AS rn
  FROM lab5.silver.slv_customers_clean
  WHERE valid_to_ts IS NULL
),
customers_with_orders AS (
  SELECT DISTINCT customer_id
  FROM lab5.silver.slv_sales_orders_clean
)
SELECT 
  ABS(HASH(c.customer_id)) AS customer_key,
  
  c.customer_id,
  
  c.customer_name,
  c.tax_id,
  c.tax_code,
  c.state,
  c.city,
  c.region,
  c.postcode,
  
  c.loyalty_segment,
  CASE c.loyalty_segment
    WHEN 0 THEN 'None'
    WHEN 1 THEN 'Bronze'
    WHEN 2 THEN 'Silver'
    WHEN 3 THEN 'Gold'
    ELSE 'Unknown'
  END AS loyalty_segment_name,
  
  c.units_purchased,
  
  CASE 
    WHEN o.customer_id IS NOT NULL THEN TRUE 
    ELSE FALSE 
  END AS has_orders,
  
  DATE(c.valid_from_ts) AS effective_date,
  
  TRUE AS is_current
  
FROM current_customers c
LEFT JOIN customers_with_orders o 
  ON c.customer_id = o.customer_id
WHERE c.rn = 1
ORDER BY c.customer_id;
```

```python
%sql
SELECT 
  'Total rows' AS metric,
  CAST(COUNT(*) AS STRING) AS value
FROM lab5.gold.dim_customers

UNION ALL

SELECT 
  'Unique customer_id' AS metric,
  CAST(COUNT(DISTINCT customer_id) AS STRING) AS value
FROM lab5.gold.dim_customers

UNION ALL

SELECT 
  'Unique customer_key' AS metric,
  CAST(COUNT(DISTINCT customer_key) AS STRING) AS value
FROM lab5.gold.dim_customers;
```

```python
%sql
CREATE OR REPLACE TABLE lab5.gold.dim_products
COMMENT 'Product dimension with ordering history for Lab 6 Gold analytics'
AS
WITH exploded_products AS (
  SELECT 
    DATE(order_ts) as order_date,
    product.id as product_id,
    product.name as product_name
  FROM lab5.silver.slv_sales_orders_clean
  LATERAL VIEW EXPLODE(ordered_products) AS product
),
product_aggregates AS (
  SELECT 
    product_id,
    product_name,
    MIN(order_date) as first_order_date,
    MAX(order_date) as last_order_date,
    COUNT(*) as times_ordered
  FROM exploded_products
  GROUP BY product_id, product_name
)
SELECT 
  ABS(HASH(product_id)) AS product_key,
  
  product_id,
  
  product_name,
  
  first_order_date,
  last_order_date,
  times_ordered
  
FROM product_aggregates
ORDER BY product_id;
```

```python
%sql
SELECT 
  'Total rows' AS metric,
  CAST(COUNT(*) AS STRING) AS value
FROM lab5.gold.dim_products

UNION ALL

SELECT 
  'Unique product_id' AS metric,
  CAST(COUNT(DISTINCT product_id) AS STRING) AS value
FROM lab5.gold.dim_products

UNION ALL

SELECT 
  'Unique product_key' AS metric,
  CAST(COUNT(DISTINCT product_key) AS STRING) AS value
FROM lab5.gold.dim_products;
```

```python
%sql
CREATE OR REPLACE TABLE lab5.gold.fct_sales_orders
COMMENT 'Sales order line items fact table with one row per product ordered'
AS
WITH exploded_orders AS (
  SELECT 
    o.order_number,
    o.customer_id,
    o.order_ts,
    product.id as product_id,
    product.name as product_name,
    product.price as price_cents,
    product.qty as quantity,
    product.curr as currency,
    product.unit as unit_of_measure,
    product.promotion_info.promo_disc as promo_disc_pct,
    product.promotion_info.promo_id as promo_id,
    ROW_NUMBER() OVER (PARTITION BY o.order_number ORDER BY product.id) as line_sequence
  FROM lab5.silver.slv_sales_orders_clean o
  LATERAL VIEW EXPLODE(o.ordered_products) AS product
)
SELECT 
  e.order_number * 1000 + e.line_sequence AS order_line_key,
  
  e.order_number,
  c.customer_key,
  p.product_key,
  d.date_key,
  
  e.order_ts,
  
  e.product_id,
  
  e.quantity,
  CAST(e.price_cents / 100.0 AS DECIMAL(10,2)) as unit_price,
  CAST((e.price_cents * e.quantity) / 100.0 AS DECIMAL(10,2)) as line_total,
  
  CAST(
    COALESCE(e.price_cents * e.quantity * e.promo_disc_pct, 0) / 100.0 
    AS DECIMAL(10,2)
  ) as promotion_discount,
  COALESCE(e.promo_id, 0) as promotion_id,
  CASE 
    WHEN e.promo_disc_pct IS NOT NULL AND e.promo_disc_pct > 0 
    THEN TRUE 
    ELSE FALSE 
  END as has_promotion,
  
  e.currency,
  e.unit_of_measure
  
FROM exploded_orders e
INNER JOIN lab5.gold.dim_customers c 
  ON e.customer_id = c.customer_id
INNER JOIN lab5.gold.dim_products p 
  ON e.product_id = p.product_id
INNER JOIN lab5.gold.dim_date d 
  ON DATE(e.order_ts) = d.full_date
ORDER BY e.order_number, e.line_sequence;
```

```python
%sql
SELECT 
  'Total fact rows' AS metric,
  CAST(COUNT(*) AS STRING) AS value
FROM lab5.gold.fct_sales_orders

UNION ALL

SELECT 
  'Distinct order_line_key' AS metric,
  CAST(COUNT(DISTINCT order_line_key) AS STRING) AS value
FROM lab5.gold.fct_sales_orders

UNION ALL

SELECT 
  'Distinct orders' AS metric,
  CAST(COUNT(DISTINCT order_number) AS STRING) AS value
FROM lab5.gold.fct_sales_orders

UNION ALL

SELECT 
  'Distinct customers' AS metric,
  CAST(COUNT(DISTINCT customer_key) AS STRING) AS value
FROM lab5.gold.fct_sales_orders

UNION ALL

SELECT 
  'Distinct products' AS metric,
  CAST(COUNT(DISTINCT product_key) AS STRING) AS value
FROM lab5.gold.fct_sales_orders;
```

## Aggregations

Customer-level and daily-level aggregates for business reporting and alerting.

```python
%sql
CREATE OR REPLACE TABLE lab5.gold.agg_customer_summary
COMMENT 'Customer-level sales summary with lifetime value metrics for customers who have placed orders'
AS
WITH customer_metrics AS (
  SELECT 
    f.customer_key,
    COUNT(DISTINCT f.order_number) as total_orders,
    COUNT(*) as total_line_items,
    SUM(f.quantity) as total_quantity,
    SUM(f.line_total) as total_revenue,
    MIN(DATE(f.order_ts)) as first_order_date,
    MAX(DATE(f.order_ts)) as last_order_date
  FROM lab5.gold.fct_sales_orders f
  GROUP BY f.customer_key
)
SELECT 
  cm.customer_key,
  c.customer_id,
  c.customer_name,
  c.state,
  c.loyalty_segment_name,
  
  cm.total_orders,
  cm.total_line_items,
  cm.total_quantity,
  CAST(cm.total_revenue AS DECIMAL(10,2)) as total_revenue,
  CAST(cm.total_revenue / cm.total_orders AS DECIMAL(10,2)) as avg_order_value,
  
  cm.first_order_date,
  cm.last_order_date,
  DATEDIFF(CURRENT_DATE(), cm.last_order_date) as days_since_last_order
  
FROM customer_metrics cm
INNER JOIN lab5.gold.dim_customers c 
  ON cm.customer_key = c.customer_key
ORDER BY cm.total_revenue DESC;
```

```python
%sql
CREATE OR REPLACE TABLE lab5.gold.agg_daily_sales
COMMENT 'Daily sales summary with order volume, revenue, and customer metrics for dashboard and alerting'
AS
WITH daily_metrics AS (
  SELECT 
    DATE(f.order_ts) as order_date,
    COUNT(DISTINCT f.order_number) as total_orders,
    COUNT(*) as total_line_items,
    SUM(f.quantity) as total_quantity,
    SUM(f.line_total) as total_revenue,
    COUNT(DISTINCT f.customer_key) as unique_customers
  FROM lab5.gold.fct_sales_orders f
  GROUP BY DATE(f.order_ts)
)
SELECT 
  dm.order_date,
  d.date_key,
  
  dm.total_orders,
  dm.total_line_items,
  dm.total_quantity,
  dm.unique_customers,
  
  CAST(dm.total_revenue AS DECIMAL(10,2)) as total_revenue,
  CAST(dm.total_revenue / dm.total_orders AS DECIMAL(10,2)) as avg_order_value
  
FROM daily_metrics dm
INNER JOIN lab5.gold.dim_date d 
  ON dm.order_date = d.full_date
ORDER BY dm.order_date;
```

```python
%sql
SELECT 
  'Total customer rows' AS metric,
  CAST(COUNT(*) AS STRING) AS value
FROM lab5.gold.agg_customer_summary

UNION ALL

SELECT 
  'Unique customer_key' AS metric,
  CAST(COUNT(DISTINCT customer_key) AS STRING) AS value
FROM lab5.gold.agg_customer_summary

UNION ALL

SELECT 
  'Unique customer_id' AS metric,
  CAST(COUNT(DISTINCT customer_id) AS STRING) AS value
FROM lab5.gold.agg_customer_summary;
```

```python
%sql
SELECT 
  'Total date rows' AS metric,
  CAST(COUNT(*) AS STRING) AS value
FROM lab5.gold.agg_daily_sales

UNION ALL

SELECT 
  'Unique order_date' AS metric,
  CAST(COUNT(DISTINCT order_date) AS STRING) AS value
FROM lab5.gold.agg_daily_sales

UNION ALL

SELECT 
  'Min order_date' AS metric,
  CAST(MIN(order_date) AS STRING) AS value
FROM lab5.gold.agg_daily_sales

UNION ALL

SELECT 
  'Max order_date' AS metric,
  CAST(MAX(order_date) AS STRING) AS value
FROM lab5.gold.agg_daily_sales;
```
