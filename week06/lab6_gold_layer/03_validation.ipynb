Final checks for the Gold layer tables and governance controls.

## Gold Layer Checks

Verifies row counts, data quality, and referential integrity.

```python
%sql
SELECT 'dim_date' as table_name, COUNT(*) as row_count 
FROM lab5.gold.dim_date

UNION ALL

SELECT 'dim_customers', COUNT(*) 
FROM lab5.gold.dim_customers

UNION ALL

SELECT 'dim_products', COUNT(*) 
FROM lab5.gold.dim_products;
```

```python
%sql
SELECT 
  COUNT(*) as total_rows,
  COUNT(DISTINCT order_line_key) as unique_keys,
  COUNT(DISTINCT order_number) as distinct_orders,
  COUNT(DISTINCT customer_key) as distinct_customers,
  COUNT(DISTINCT product_key) as distinct_products,
  COUNT(customer_key) as non_null_customer_keys,
  COUNT(product_key) as non_null_product_keys,
  COUNT(date_key) as non_null_date_keys
FROM lab5.gold.fct_sales_orders;
```

```python
%sql
SELECT 
  'fct_sales_orders' as source, 
  COUNT(DISTINCT order_number) as orders, 
  ROUND(SUM(line_total), 2) as revenue
FROM lab5.gold.fct_sales_orders

UNION ALL

SELECT 
  'agg_customer_summary', 
  SUM(total_orders), 
  ROUND(SUM(total_revenue), 2)
FROM lab5.gold.agg_customer_summary

UNION ALL

SELECT 
  'agg_daily_sales', 
  SUM(total_orders), 
  ROUND(SUM(total_revenue), 2)
FROM lab5.gold.agg_daily_sales;
```

## Governance Validation

Verifies RLS and CLS are working correctly.

```python
%sql
SELECT
    state,
    COUNT(*) AS customer_count
FROM lab5.gold.dim_customers
GROUP BY state
ORDER BY state;
```
